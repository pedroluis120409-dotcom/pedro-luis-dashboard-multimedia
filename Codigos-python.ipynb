# Widget 2 Password Creator

```python
import random
import string
import ipywidgets as widgets
from IPython.display import display

# 1. Dados Base
PALAVRAS = ["Tigre", "Cometa", "Pixel", "Fenix", "Oceano", "Codigo", "Saturno", "Raio", "Falcao", "Prisma"]
ADJETIVOS = ["Azul", "Veloz", "Solar", "Epico", "Digital", "Forte", "Cosmico", "Nobre"]
SIMBOLOS = ["!", "@", "#", "$", "%", "&", "*"]

DICAS = [
    "Usa um gestor de palavras-passe para não teres de decorar dezenas de códigos.",
    "Ativa sempre a autenticação de dois fatores (2FA) nas tuas contas.",
    "Nunca reutilizes a mesma palavra-passe no email e nas redes sociais.",
    "As frases-passe combinam alta segurança com facilidade de memorização."
]

# 2. Componentes Nativos (100% Python / ipywidgets)
titulo = widgets.Label(value="🔐 GERADOR DE PASSWORD FORTE")

modo_toggle = widgets.ToggleButtons(
    options=[('🎲 Aleatória', 'random'), ('🧠 Frase-Passe', 'phrase')],
    description='Modo:'
)

len_slider = widgets.IntSlider(value=14, min=6, max=24, description='Tamanho:')
chk_upper = widgets.Checkbox(value=True, description='Maiúsculas (A-Z)')
chk_numbers = widgets.Checkbox(value=True, description='Números (0-9)')
chk_symbols = widgets.Checkbox(value=True, description='Símbolos (!@#$%)')
chk_no_amb = widgets.Checkbox(value=False, description='Sem ambíguos (0/O/1/l/I)')

opcoes_aleatorias = widgets.VBox([
    len_slider,
    widgets.HBox([chk_upper, chk_numbers]),
    widgets.HBox([chk_symbols, chk_no_amb])
])

info_frase = widgets.Label(
    value="Gera combinações memoráveis (ex: Tigre#Azul42!).",
    layout=widgets.Layout(display='none')
)

btn_gerar = widgets.Button(
    description='Gerar Nova Palavra-Passe',
    button_style='primary',
    icon='key',
    layout=widgets.Layout(width='100%')
)

# Área de Resultados Nativa
campo_password = widgets.Text(
    value='',
    description='Password:',
    disabled=False,
    layout=widgets.Layout(width='100%')
)

barra_forca = widgets.IntProgress(
    value=100, min=0, max=100,
    description='Força:',
    bar_style='success',
    layout=widgets.Layout(width='100%')
)

lbl_estado = widgets.Label(value="Força: --- | Quebra: ---")
lbl_dica = widgets.Label(value="🧠 Nota IA: ---")

# 3. Lógica em Python Puro
def gerar_password(_=None):
    if modo_toggle.value == 'phrase':
        p1 = random.choice(PALAVRAS)
        p2 = random.choice(ADJETIVOS)
        s1 = random.choice(SIMBOLOS)
        num = random.randint(10, 98)
        pwd = f"{p1}{s1}{p2}{num}!"
        score = 3
    else:
        tamanho = len_slider.value
        charset = string.ascii_lowercase
        if chk_upper.value:
            charset += string.ascii_uppercase
        if chk_numbers.value:
            charset += string.digits
        if chk_symbols.value:
            charset += "!@#$%&*?"
        if chk_no_amb.value:
            for c in "0O1lI":
                charset = charset.replace(c, "")

        pwd = "".join(random.choice(charset) for _ in range(tamanho))

        if tamanho >= 12 and (chk_upper.value or chk_numbers.value) and chk_symbols.value:
            score = 3
        elif tamanho >= 8 and (chk_upper.value or chk_numbers.value or chk_symbols.value):
            score = 2
        else:
            score = 1

    # Atualizar widgets nativos
    campo_password.value = pwd

    if score == 1:
        barra_forca.value = 30
        barra_forca.bar_style = 'danger'
        lbl_estado.value = "Força: Fraca | Tempo de quebra: Segundos"
    elif score == 2:
        barra_forca.value = 65
        barra_forca.bar_style = 'warning'
        lbl_estado.value = "Força: Média | Tempo de quebra: Alguns meses"
    else:
        barra_forca.value = 100
        barra_forca.bar_style = 'success'
        lbl_estado.value = "Força: Invencível | Tempo de quebra: +500 anos"

    lbl_dica.value = f"🧠 Nota IA: {random.choice(DICAS)}"

def alternar_modo(change):
    if change['new'] == 'random':
        opcoes_aleatorias.layout.display = 'flex'
        info_frase.layout.display = 'none'
    else:
        opcoes_aleatorias.layout.display = 'none'
        info_frase.layout.display = 'block'
    gerar_password()

# 4. Eventos
modo_toggle.observe(alternar_modo, names='value')
len_slider.observe(gerar_password, names='value')
chk_upper.observe(gerar_password, names='value')
chk_numbers.observe(gerar_password, names='value')
chk_symbols.observe(gerar_password, names='value')
chk_no_amb.observe(gerar_password, names='value')
btn_gerar.on_click(gerar_password)

# 5. Layout Final
interface = widgets.VBox([
    titulo,
    modo_toggle,
    opcoes_aleatorias,
    info_frase,
    btn_gerar,
    campo_password,
    barra_forca,
    lbl_estado,
    lbl_dica
], layout=widgets.Layout(width='480px', padding='12px', border='1px solid #ccc'))

display(interface)
gerar_password()
```
